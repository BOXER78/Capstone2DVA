# 01 Extraction

Load the raw student mental health & burnout dataset, inspect its structure, and confirm it is suitable for the project pipeline.

**Dataset:** `student_mental_health_burnout_1M.csv`  
**Rows:** ~1,000,000  
**Columns:** 20  
**Granularity:** One row per student observation

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().resolve().name == 'notebooks' else Path.cwd().resolve()
RAW_PATH = PROJECT_ROOT / 'data/raw/student_mental_health_burnout_1M.csv'

print(f'Project root: {PROJECT_ROOT}')
print(f'Raw data path: {RAW_PATH}')
print(f'File exists: {RAW_PATH.exists()}')
```

## 1.1 Load the Dataset

```python
df = pd.read_csv(RAW_PATH)
print(f'Shape: {df.shape}')
df.head(10)
```

## 1.2 Schema Inspection

```python
print(f'Rows   : {len(df):,}')
print(f'Columns: {len(df.columns)}')
print()
df.info()
```

## 1.3 Column Names

```python
for i, col in enumerate(df.columns, 1):
    print(f'{i:>2}. {col}')
```

## 1.4 Descriptive Statistics

```python
df.describe(include='all').T
```

## 1.5 Missing Values

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_report = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_report[missing_report['missing_count'] > 0]
```

## 1.6 Categorical Column Distributions

```python
cat_cols = df.select_dtypes(include='object').columns.tolist()
print(f'Categorical columns: {cat_cols}')
for col in cat_cols:
    print(f'\n--- {col} ---')
    print(df[col].value_counts())
```

## 1.7 Numeric Range Check

```python
num_cols = df.select_dtypes(include='number').columns.tolist()
df[num_cols].agg(['min', 'max', 'mean', 'std']).T
```

## 1.8 Duplicate Check

```python
dup_count = df.duplicated().sum()
print(f'Duplicate rows: {dup_count:,} ({dup_count / len(df) * 100:.2f}%)')
```
